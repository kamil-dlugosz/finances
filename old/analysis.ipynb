```python
import pandas as pd

csv_paths = [
    "2025_1.csv",
    "2025_0.csv"
]
dfs = []
for f in csv_paths:
    dfs.append(pd.read_csv(
        f,
        sep=";",                # semicolon separator
        quotechar='"',          # handle quoted fields properly
        decimal=",",            # interpret ',' as decimal
        engine="python",        # more forgiving parser
        dtype=str               # read everything as string first to avoid type issues
    ))
df = pd.concat(dfs, ignore_index=True)
df["Kwota operacji"] = (
    df["Kwota operacji"]
    .astype(str)
    .str.replace(" ", "", regex=False)   # remove spaces
    .str.replace(",", ".", regex=False)  # handle comma as decimal if needed
    .astype(float)
)
```

```python
df = df[df["Kwota operacji"] < 0]
df
```

```python
df.groupby("Kategoria")["Kwota operacji"].sum()
```

```python
df[df["Kategoria"] == 'Inne']
```

```python
category_sums = df.groupby("Kategoria")["Kwota operacji"].sum().reset_index()
category_sums = category_sums.sort_values("Kwota operacji", ascending=False)

df_merged = df.merge(category_sums, on="Kategoria", suffixes=("", "_category_sum"))

df_sorted = df_merged.sort_values(
    by=["Kwota operacji_category_sum", "Kwota operacji"], 
    ascending=[True, True]
)
df_sorted
```

```python
df_sorted["kat"] = ""

cols_first = [
    "Data księgowania",
    "Nadawca / Odbiorca",
    "Adres nadawcy / odbiorcy",
    "Tytułem",
    "Kwota operacji",
    "Kategoria",
    "kat"
]
other_cols = [c for c in df_sorted.columns if c not in cols_first]

df_sorted = df_sorted[cols_first + other_cols]
df_sorted
```

```python
mask = df_sorted["Tytułem"].str.contains("bilet", case=False, na=False) & \
       df_sorted["Tytułem"].str.contains("warszawa", case=False, na=False)
df_sorted.loc[mask, "kat"] = "Warszawa bilety"
df_sorted[mask]
```

```python
mask = df_sorted["Adres nadawcy / odbiorcy"].str.contains("kokonki", case=False, na=False) #& \
       # df_sorted["Tytułem"].str.contains("warszawa", case=False, na=False)
df_sorted.loc[mask, "kat"] = "włóczki"
df_sorted[mask]
```

```python
mask = df_sorted["Adres nadawcy / odbiorcy"].str.contains("znanylekarz", case=False, na=False) #& \
       # df_sorted["Tytułem"].str.contains("warszawa", case=False, na=False)
df_sorted.loc[mask, "kat"] = "zdrowie"
df_sorted[mask]
```
